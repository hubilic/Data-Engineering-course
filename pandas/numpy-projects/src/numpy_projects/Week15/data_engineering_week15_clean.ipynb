```python
CSV_TEXT = """sale_id,date,branch,product,unit_price,quantity,discount,payment
1,2026-08-24,Central,Coffee,3.50,2,0,Card
2,2026-08-24,Central,Tea,2.50,3,0,Cash
3,2026-08-24,Central,Cake,4.00,1,0.10,Card
4,2026-08-24,Station,Sandwich,6.00,2,0,Card
5,2026-08-24,Station,coffee,3.50,4,0.10,Cash
6,2026-08-25,Central, Coffee ,3.50,3,,Card
7,2026-08-25,Central,Sandwich,6.00,1,0.10,Card
8,2026-08-25,Station,Tea,2.50,5,0,Cash
9,2026-08-25,Station,Cake,4.00,2,0.20,Card
10,2026-08-25,Station,Coffee,3.50,1,0,
11,2026-08-26,Central,Tea,2.50,2,0,Cash
12,2026-08-26,Central,Cake,4.00,3,0,Card
13,2026-08-26,Station,Sandwich,6.00,3,0.10,Card
14,2026-08-26,Station,Coffee,3.50,2,0,Card
15,2026-08-26,Central,Coffee,3.50,4,0.20,Cash
16,2026-08-27,Central,Sandwich,6.00,2,0,Card
17,2026-08-27,Central,Cake,4.00,2,,Cash
18,2026-08-27,Station,Tea,2.50,4,0.10,Card
19,2026-08-27,Station,Coffee,3.50,3,0,Card
20,2026-08-27,Central,Tea,2.50,1,0,Card
5,2026-08-24,Station,coffee,3.50,4,0.10,Cash
"""

raw = pd.read_csv(StringIO(CSV_TEXT))

menu = pd.DataFrame({
    "product": ["Coffee", "Tea", "Cake", "Sandwich"],
    "category": ["Drink", "Drink", "Food", "Food"],
    "unit_cost": [1.00, 0.50, 1.50, 2.80],
})

raw.head()
```

```python
sales = raw.drop_duplicates().copy()
sales["product"] = sales["product"].str.strip().str.title()
sales["discount"] = sales["discount"].fillna(0)
sales["payment"] = sales["payment"].fillna("Unknown")
assert sales["sale_id"].is_unique, "Conflicting sale IDs need investigation."
display(sales.head())
```

```python
display(sales["quantity"] >= 3)
sales[sales["quantity"] >= 3]
```

```python
mask = (sales["branch"] == "Central") & (sales["quantity"] >= 3)
sales.loc[mask, ["sale_id", "product", "quantity"]]
```

```python
# TODO: Find Station coffee sales with quantity at least 3. Show sale_id, product, and quantity.
mask = ((sales["product"] == "Coffee") & (sales["branch"] == "Station") & (sales["quantity"] >= 3))
sales.loc[mask, ["sale_id", "product", "quantity"]]
```

```python
# TODO: Find Tea or Cake sales paid by card.

mask = (((sales["product"] == "Tea") | (sales["product"] == "Cake")) & (sales["payment"] == "Card"))
display(sales[mask])
```

```python
sales["gross_revenue"] = sales["unit_price"] * sales["quantity"]
sales["revenue"] = (sales["gross_revenue"] * (1 - sales["discount"])).round(2)
sales["discount_amount"] = (sales["gross_revenue"] - sales["revenue"]).round(2)
sales.loc[sales["sale_id"] == 5, ["gross_revenue", "discount_amount", "revenue"]]
```

```python
# TODO: Calculate total units, gross revenue, revenue, and discount amount. Show the three highest-revenue individual sales.
sales[["quantity", "gross_revenue", "revenue", "discount_amount"]].sum()
display(sales.sort_values("revenue", ascending=False).head(3))


```

```python
# TODO: Find discounted sales with revenue of at least 10. Show sale_id, product, discount, and revenue. Keep sales unchanged.
mask = ((sales["discount"] > 0) & (sales["revenue"] >= 10))
sales.loc[mask, ["sale_id", "product", "discount", "revenue"]]
```

```python
product_summary = sales.groupby("product", as_index=False).agg(
    sales_count=("sale_id", "size"),
    units=("quantity", "sum"),
    revenue=("revenue", "sum"),
)
product_summary = product_summary.sort_values("revenue", ascending=False)
product_summary
```

```python
# TODO: Group by branch. Calculate sales count, units, total revenue, and mean revenue per sale. Rank branches by total revenue.
branch_summary = sales.groupby("branch", as_index= False).agg(
    sales_count=("sale_id", "size"),
    units=("quantity", "sum"),
    revenue=("revenue", "sum"),
    avg_transaction=("revenue", "mean")
)
branch_summary = branch_summary.sort_values("revenue", ascending=False)
branch_summary
```

```python
# TODO: Group by branch and product. Show sales count, units, and revenue, sorted by branch and descending revenue.
branch_product_summary = sales.groupby(["branch", "product"], as_index=False).agg(
    sales_count=("sale_id", "size"),
    units=("quantity", "sum"),
    revenue=("revenue", "sum"),
)
branch_product_summary = branch_product_summary.sort_values(["branch", "revenue"], ascending=[True, False])
branch_product_summary

```

```python
display(menu)
enriched = sales.merge(menu, on="product", how="left", validate="many_to_one")
assert len(enriched) == len(sales)
assert enriched["unit_cost"].notna().all(), "Some products are missing ingredient costs."
enriched["ingredient_cost"] = (enriched["quantity"] * enriched["unit_cost"]).round(2)
enriched["contribution"] = (enriched["revenue"] - enriched["ingredient_cost"]).round(2)
enriched.head()
```

```python
# TODO: Rank products by total contribution. Compare this with the revenue ranking. Calculate total contribution across all sales.
ranking = enriched.groupby("product", as_index=False).agg(
    revenue=("revenue", "sum"),
    contribution=("contribution", "sum"),
    ingedient_cost=("ingredient_cost", "sum")
)
ranking = ranking.sort_values("contribution", ascending=False)
com_rank = ranking.copy()

com_rank["revenue_rank"] = (
    com_rank["revenue"]
    .rank(ascending=False, method="min")
    .astype(int)
)

com_rank["contribution_rank"] = (
    com_rank["contribution"]
    .rank(ascending=False, method="min")
    .astype(int)
)

com_rank = com_rank.sort_values("revenue_rank")
display(com_rank)
display(ranking)


com_rank = enriched["contribution"].sum()
print(com_rank)
```

```python
# TODO: Remove Tea from a copy of menu. Left-merge sales with this lookup using indicator=True and find unmatched sales. Compare with an inner merge. Keep menu and enriched unchanged.
menu_without_tea = menu.loc[menu["product"] != "Tea"].copy()
left_res = sales.merge(
    menu_without_tea,
    on="product",
    how="left",
    indicator=True,
    validate="many_to_one"
)

unmatched = left_res.loc[left_res["_merge"] == "left_only"]

inner_res = sales.merge(
    menu_without_tea,
    on="product",
    how="inner",
)

comparison = pd.DataFrame({
    "rows": [
        len(sales),
        len(left_res),
        len(inner_res)
    ],

    "revenue": [
        sales["revenue"].sum(),
        left_res["revenue"].sum(),
        inner_res["revenue"].sum()
    ]
},
index=["Original sales", "Left merge", "innner merge"]

)

display(comparison)



```
